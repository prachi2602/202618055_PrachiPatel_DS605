## Task 2 - Vectorized Arithmetic and Linear Algebra


### 1. Create Two Matrices

```python
matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)
```

### 2. Matrix Addition

```python
addition = matrix_a + matrix_b

print("Matrix Addition:")
print(addition)
```

### 3. Element-wise Multiplication



```python
element_multiplication = matrix_a * matrix_b

print("Element-wise Multiplication:")
print(element_multiplication)
```

### 4. Matrix Multiplication


```python
matrix_multiplication = matrix_a @ matrix_b

print("Matrix Multiplication:")
print(matrix_multiplication)
```

### 5.Transpose of Matrix A

```python
transpose_a = matrix_a.T

print("Original Matrix A:")
print(matrix_a)

print("\nTranspose of Matrix A:")
print(transpose_a)
```

### 6.Determinant of Matrix A

```python
determinant_a = np.linalg.det(matrix_a)

print("Determinant of Matrix A:", determinant_a)
```

### 7.Inverse of Matrix A

```python
inverse_a = np.linalg.inv(matrix_a)

print("Inverse of Matrix A:")
print(inverse_a)
```

### 8. Verification of the Inverse

```python
verification_matrix = matrix_a @ inverse_a

print("Matrix A @ Inverse of A:")
print(verification_matrix)

identity_matrix = np.eye(2)

print("\nIdentity Matrix:")
print(identity_matrix)

print("\nVerification using np.allclose():")
print(np.allclose(verification_matrix, identity_matrix))
```

## Task 5 - Filtering and Querying

### Male Passengers Older Than 50

```python
male_over_50 = titanic_df[
    (titanic_df["Sex"] == "male") &
    (titanic_df["Age"] > 50)
]

print("Male passengers older than 50:", len(male_over_50))
```

### Female First-Class Passengers


```python
female_first_class = titanic_df[
    (titanic_df["Sex"] == "female") &
    (titanic_df["Pclass"] == 1)
]

female_first_count = len(female_first_class)

female_first_survival_rate = (
    female_first_class["Survived"].mean() * 100
)

print("Female first-class passengers:", female_first_count)
print("Survival percentage:", round(female_first_survival_rate, 2), "%")
```

### Passengers Aged 20-40 with Above-Median Fare Who Survived


```python
overall_fare_median = titanic_df["Fare"].median()

age_fare_survived = titanic_df[
    (titanic_df["Age"].between(20, 40)) &
    (titanic_df["Fare"] > overall_fare_median) &
    (titanic_df["Survived"] == 1)
]

print("Overall median Fare:", overall_fare_median)
print("Number of passengers:", len(age_fare_survived))
```

### Passengers Travelling Alone, Under 30, Who Did Not Survive


```python
alone_under_30_not_survived = titanic_df[
    (titanic_df["SibSp"] == 0) &
    (titanic_df["Parch"] == 0) &
    (titanic_df["Age"] < 30) &
    (titanic_df["Survived"] == 0)
]

print(
    "Travelling alone, age under 30, and did not survive:",
    len(alone_under_30_not_survived)
)
```

### Southampton Passengers in Class 2 or 3 with Above-Median Fare


```python
southampton_median_fare = titanic_df.loc[
    titanic_df["Embarked"] == "S",
    "Fare"
].median()

southampton_filtered = titanic_df[
    (titanic_df["Embarked"] == "S") &
    (titanic_df["Pclass"].isin([2, 3])) &
    (titanic_df["Fare"] > southampton_median_fare)
]

print("Southampton median Fare:", southampton_median_fare)
print("Number of passengers:", len(southampton_filtered))
```

## Task 6 - `groupby()` and Aggregation


### Survival Rate by Sex


```python
survival_by_sex = titanic_df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Sex (%):")
print(survival_by_sex.round(2))
```

### Survival Rate by Passenger Class


```python
survival_by_pclass = titanic_df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Pclass (%):")
print(survival_by_pclass.round(2))
```

### Average Age and Fare by Pclass


```python
avg_age_fare_by_pclass = titanic_df.groupby("Pclass")[["Age", "Fare"]].mean()

print("Average Age and Fare by Pclass:")
print(avg_age_fare_by_pclass.round(2))
```

### Passenger Count and Survival Rate by Sex and Pclass



```python
sex_pclass_summary = titanic_df.groupby(
    ["Sex", "Pclass"]
).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

print("Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_pclass_summary.round(2))
```

### Passenger Count, Average Fare, and Survival Rate by Embarked


```python
embarked_summary = titanic_df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

print("Summary by Embarked:")
print(embarked_summary.round(2))
```

### Filling Missing Age Values with Mean


```python
mean_age_df = titanic_df.copy()

print("Missing Age values before:",
      mean_age_df["Age"].isnull().sum())

mean_age = mean_age_df["Age"].mean()

print("Mean Age:", round(mean_age, 2))

mean_age_df["Age"] = mean_age_df["Age"].fillna(mean_age)

print("Missing Age values after:",
      mean_age_df["Age"].isnull().sum())
```

### Mean Imputation

```python
age_mean_df = titanic_df.copy()

age_mean = age_mean_df["Age"].mean()

age_mean_df["Age"] = age_mean_df["Age"].fillna(age_mean)

print("Mean used:", round(age_mean, 2))
print("Missing Age values:", age_mean_df["Age"].isnull().sum())
```

### Median Imputation


```python
age_median_df = titanic_df.copy()

age_median = age_median_df["Age"].median()

age_median_df["Age"] = age_median_df["Age"].fillna(age_median)

print("Median used:", age_median)
print("Missing Age values:", age_median_df["Age"].isnull().sum())
```

### Mode Imputation


```python
age_mode_df = titanic_df.copy()

age_mode = age_mode_df["Age"].mode()[0]

age_mode_df["Age"] = age_mode_df["Age"].fillna(age_mode)

print("Mode used:", age_mode)
print("Missing Age values:", age_mode_df["Age"].isnull().sum())
```

### Random Value Imputation


```python
random_age_df = titanic_df.copy()

np.random.seed(42)

missing_age_count = random_age_df["Age"].isnull().sum()

random_values = np.random.choice(
    random_age_df["Age"].dropna(),
    size=missing_age_count
)

random_age_df.loc[
    random_age_df["Age"].isnull(),
    "Age"
] = random_values

print("Missing Age values after random imputation:",
      random_age_df["Age"].isnull().sum())
```

### Detecting Fare Outliers using the IQR Method


```python
Q1 = titanic_df["Fare"].quantile(0.25)
Q3 = titanic_df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
fare_outliers = titanic_df[
    (titanic_df["Fare"] < lower_bound) |
    (titanic_df["Fare"] > upper_bound)
]

print("Number of Fare Outliers:", len(fare_outliers))
```

## Task 8 - Feature Engineering and Pivot Table


### Create FamilySize


```python
titanic_df["FamilySize"] = (
    titanic_df["SibSp"] +
    titanic_df["Parch"] +
    1
)

titanic_df[["SibSp", "Parch", "FamilySize"]].head()
```

### Create IsAlone


```python
titanic_df["IsAlone"] = np.where(
    titanic_df["FamilySize"] == 1,
    1,
    0
)

titanic_df[["FamilySize", "IsAlone"]].head(10)
```

### Survival Pivot Table by Sex and Pclass


```python
survival_pivot = pd.pivot_table(
    titanic_df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

survival_pivot
```

### Highest Survival Group

```python
highest_group = survival_pivot.stack().idxmax()
highest_rate = survival_pivot.stack().max() * 100

print("Highest Survival Group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", round(highest_rate, 2), "%")
```

### Lowest Survival Group

```python
lowest_group = survival_pivot.stack().idxmin()
lowest_rate = survival_pivot.stack().min() * 100

print("Lowest Survival Group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", round(lowest_rate, 2), "%")
```

### Final Observations

1. Female passengers had a noticeably higher survival rate than male passengers.

2. First-class passengers had a higher survival rate compared with passengers in the lower passenger classes.

3. Passenger class showed a negative relationship with survival because a higher numerical Pclass value represents a lower passenger class.

4. Fare showed a positive relationship with survival, suggesting that passengers who paid higher fares generally had a better survival outcome.

5. The Age vs Fare scatter plot shows that passengers of many different ages were present in both the survived and non-survived groups, so Age alone does not clearly separate survival outcomes.

6. Family-related features such as FamilySize and IsAlone provide additional information about whether passengers were travelling alone or with family.

7. The correlation heatmap shows that some variables are related to each other, but correlation only measures association and does not prove that one variable caused another.
