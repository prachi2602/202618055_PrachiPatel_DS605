## Task 2 - Missing Values, Leakage, and Outliers

### Missing Values

```python
missing_values = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_values.sort_values("Missing Percentage", ascending=False)
```

### Columns with Very High Missingness

```python
high_missing = missing_values[
    missing_values["Missing Percentage"] > 80
]

high_missing
```

### Remove High-Missingness Column

`company` is removed because 94.31% of its values are missing, making reliable imputation unsuitable.

```python
df_clean = df.copy()

if "company" in df_clean.columns:
    df_clean = df_clean.drop(columns=["company"])
```

### Remove Leakage Columns

```python
leakage_columns = [
    "reservation_status",
    "reservation_status_date"
]

df_clean = df_clean.drop(
    columns=[col for col in leakage_columns if col in df_clean.columns]
)
```

### Selected Numerical Features for Outlier Detection

```python
outlier_columns = [
    "lead_time",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies",
    "adr"
]
```

### Boxplots

```python
for column in outlier_columns:
    plt.figure(figsize=(8, 3))
    plt.boxplot(df_clean[column].dropna(), orientation="horizontal")
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.show()
```

### Extreme Outliers Using IQR

```python
outlier_summary = []

for column in outlier_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR

    extreme_count = (
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ).sum()

    outlier_summary.append([
        column,
        lower_bound,
        upper_bound,
        extreme_count
    ])

outlier_summary_df = pd.DataFrame(
    outlier_summary,
    columns=[
        "Feature",
        "Lower Bound",
        "Upper Bound",
        "Extreme Outliers"
    ]
)

outlier_summary_df
```

### Remove Clear Extreme Outliers

```python
rows_before = len(df_clean)

extreme_outlier_mask = (
    (df_clean["adults"] > 10) |
    (df_clean["children"] > 3) |
    (df_clean["babies"] > 2) |
    (df_clean["adr"] > 1000) |
    (df_clean["adr"] < 0) |
    (df_clean["stays_in_weekend_nights"] > 10) |
    (df_clean["stays_in_week_nights"] > 20)
)

df_clean = df_clean[~extreme_outlier_mask].copy()

rows_after = len(df_clean)
rows_removed = rows_before - rows_after

print("Rows before outlier removal:", rows_before)
print("Rows after outlier removal:", rows_after)
print("Rows removed:", rows_removed)
```

## Task 3 - Create Two Preprocessing Pipelines

### Define Features and Target

```python
X = df_clean.drop(columns=["is_canceled"])
y = df_clean["is_canceled"]
```

### Identify Numerical and Categorical Columns

```python
numerical_columns = X.select_dtypes(include=np.number).columns.tolist()
categorical_columns = X.select_dtypes(exclude=np.number).columns.tolist()

print("Number of numerical columns:", len(numerical_columns))
print("Number of categorical columns:", len(categorical_columns))
```

### Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

### Import Preprocessing Tools

```python
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

### Numerical Pipeline A - StandardScaler

```python
numerical_pipeline_A = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])
```

### Numerical Pipeline B - MinMaxScaler

```python
numerical_pipeline_B = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])
```

### Categorical Preprocessing Pipeline

```python
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

### Preprocessor A

```python
preprocessor_A = ColumnTransformer(transformers=[
    ("num", numerical_pipeline_A, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns)
])
```

### Preprocessor B

```python
preprocessor_B = ColumnTransformer(transformers=[
    ("num", numerical_pipeline_B, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns)
])
```

### Pipeline A and Pipeline B

```python
pipeline_A = Pipeline(steps=[
    ("preprocessor", preprocessor_A)
])

pipeline_B = Pipeline(steps=[
    ("preprocessor", preprocessor_B)
])
```

# Part B - Model Training and Performance Evaluation
## Task 4 - Train Two Classification Models
### Import Classification Models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

### Logistic Regression with Pipeline A

```python
logistic_pipeline_A = Pipeline(steps=[
    ("preprocessor", preprocessor_A),
    ("model", LogisticRegression(max_iter=1000))
])

logistic_pipeline_A.fit(X_train, y_train)
```

### Logistic Regression with Pipeline B

```python
logistic_pipeline_B = Pipeline(steps=[
    ("preprocessor", preprocessor_B),
    ("model", LogisticRegression(max_iter=1000))
])

logistic_pipeline_B.fit(X_train, y_train)
```

### Decision Tree with Pipeline A

```python
tree_pipeline_A = Pipeline(steps=[
    ("preprocessor", preprocessor_A),
    ("model", DecisionTreeClassifier(random_state=42))
])

tree_pipeline_A.fit(X_train, y_train)
```

### Decision Tree with Pipeline B

```python
tree_pipeline_B = Pipeline(steps=[
    ("preprocessor", preprocessor_B),
    ("model", DecisionTreeClassifier(random_state=42))
])

tree_pipeline_B.fit(X_train, y_train)
```

### Four Trained Model-Pipeline Combinations

```python
trained_models = {
    "Logistic Regression - StandardScaler": logistic_pipeline_A,
    "Logistic Regression - MinMaxScaler": logistic_pipeline_B,
    "Decision Tree - StandardScaler": tree_pipeline_A,
    "Decision Tree - MinMaxScaler": tree_pipeline_B
}

print("Total trained model-pipeline combinations:", len(trained_models))
```

## Task 5 - Evaluate and Compare the Four Results

### Import Evaluation Metrics

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

### Evaluate All Four Models

```python
results = []

for name, model in trained_models.items():
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred)
    recall = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)

    train_test_difference = train_accuracy - test_accuracy

    results.append({
        "Model-Pipeline": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "Train-Test Difference": train_test_difference
    })
```

### Final Comparison Table

```python
comparison_df = pd.DataFrame(results)

comparison_df.round(4)
```

### Best Logistic Regression Model

```python
logistic_results = comparison_df[
    comparison_df["Model-Pipeline"].str.contains("Logistic Regression")
]

logistic_results
```

```python
best_logistic_name = logistic_results.loc[
    logistic_results["F1-Score"].idxmax(),
    "Model-Pipeline"
]

best_logistic_model = trained_models[best_logistic_name]

print("Best Logistic Regression:", best_logistic_name)
```

### Best Decision Tree Model

```python
tree_results = comparison_df[
    comparison_df["Model-Pipeline"].str.contains("Decision Tree")
]

tree_results
```

```python
best_tree_name = tree_results.loc[
    tree_results["F1-Score"].idxmax(),
    "Model-Pipeline"
]

best_tree_model = trained_models[best_tree_name]

print("Best Decision Tree:", best_tree_name)
```

### Confusion Matrix - Best Logistic Regression

```python
y_pred_best_logistic = best_logistic_model.predict(X_test)

cm_logistic = confusion_matrix(y_test, y_pred_best_logistic)

ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Canceled", "Canceled"]
).plot()

plt.title(f"Confusion Matrix - {best_logistic_name}")
plt.savefig(
    "figures/logistic_regression_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

### Confusion Matrix - Best Decision Tree

```python
y_pred_best_tree = best_tree_model.predict(X_test)

cm_tree = confusion_matrix(y_test, y_pred_best_tree)

ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Not Canceled", "Canceled"]
).plot()

plt.title(f"Confusion Matrix - {best_tree_name}")
plt.savefig(
    "figures/decision_tree_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

### Train-Test Performance Difference

```python
comparison_df[
    ["Model-Pipeline", "Training Accuracy", "Testing Accuracy", "Train-Test Difference"]
].round(4)
```

### Overfitting Observation

A larger difference between training and testing accuracy indicates possible overfitting. Models with a small train-test difference show better generalization to unseen data.

### Final Observations

1. Among all four combinations, the **Decision Tree with StandardScaler** gave the best overall performance, with around **86.24% testing accuracy** and an **F1-score of 81.53%**.

2. In Logistic Regression, **StandardScaler performed slightly better than MinMaxScaler**. The difference was small, but StandardScaler gave a little higher testing accuracy and F1-score.

3. For the Decision Tree, changing from StandardScaler to MinMaxScaler did not make much difference. The results were almost the same in both cases.

4. The Decision Tree was better at identifying canceled bookings because its recall was around **82%**, while the best Logistic Regression model had a recall of around **67%**.

5. The Decision Tree also showed signs of overfitting because its training accuracy was very high at about **99.62%**, while the testing accuracy was only about **86.24%**. Logistic Regression had a much smaller difference between training and testing accuracy.
